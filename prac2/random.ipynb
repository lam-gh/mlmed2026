```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np 
import cv2

pic = Image.open("../data/hc18/training_set/000_HC_Annotation.png")
pix = np.array(pic)

print(pix.shape)

contours, _ = cv2.findContours(pix, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
cv2.drawContours(pix, contours, -1, 255, -1)


plt.imshow(pix, cmap='gray')
```
